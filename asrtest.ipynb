```python
import torch
from qwen_asr import Qwen3ASRModel

model = Qwen3ASRModel.from_pretrained(
    "Qwen/Qwen3-ASR-1.7B",
    dtype=torch.bfloat16,
    device_map="cuda:0",
    # attn_implementation="flash_attention_2",
    max_inference_batch_size=32, # Batch size limit for inference. -1 means unlimited. Smaller values can help avoid OOM.
    max_new_tokens=256, # Maximum number of tokens to generate. Set a larger value for long audio input.
    forced_aligner="Qwen/Qwen3-ForcedAligner-0.6B",
    forced_aligner_kwargs=dict(
        dtype=torch.bfloat16,
        device_map="cuda:0",
        #attn_implementation="flash_attention_2",
    ),
)

```

```python
file_path= "/content/drive/MyDrive/01 - Esraa - Intro.mp3"
```

```python
import google.colab.ai as ai

results = model.transcribe(
    audio=[
      file_path
    ],
    language=None, # This parameter might influence the transcription output directly in some models, but Qwen3-ASR seems to transcribe to the detected language.
    return_time_stamps=True,
)

for r in results:
    arabic_text = r.text
    print(f"Detected Language: {r.language}")
    print(f"Transcribed Text (Arabic): {arabic_text}")
    print(f"Time Stamp: {r.time_stamps[0]}")

    # Use google.colab.ai for translation
    translation_prompt = f"Translate the following Arabic text to English: {arabic_text}"
    english_translation = ai.generate_text(translation_prompt, model_name='google/gemini-2.5-flash')
    print(f"English Translation: {english_translation}")
    print("\n" + "-"*50 + "\n") # Separator for multiple results, if any
```
